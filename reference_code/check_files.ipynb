```python
import os
import re
```

```python
mouse = "JL117"
root = rf"Z:\People\JenLi\Imaging_Data\L23_L5_structural_imaging\Analyzed_Data\{mouse}"

# check fov folders
fovs = [f for f in os.listdir(root) if f.startswith('FOV')]
print(f'{len(fovs)} FOV folders: {sorted(fovs)}')
if len(fovs) == 0:
    print('no FOVs found, check your path')
```

```python
JL117_days = ['250211', '250212', '250213', '250214', '250215', '250216', '250217', '250218', '250219', '250220', '250221', '250222', '250224', '250225']
JL122_days = ['250402', '250403', '250404', '250405', '250406', '250407', '250408', '250409', '250410', '250411', '250412', '250413', '250414', '250415']
JL123_days = ['250402', '250403', '250404', '250405', '250406', '250407', '250408', '250409', '250410', '250411', '250412', '250413', '250414', '250415']
```

```python
for fov in sorted(fovs):
    fov_path = os.path.join(root, fov)
    subfolders = os.listdir(fov_path)
    temp = []
    i = 0
    j = 0

    while i < len(JL117_days) and j < len(subfolders):
        if JL117_days[i] == subfolders[j]:
            temp.append(subfolders[j])
            i += 1
            j += 1
        else:
            temp.append("missing")
            i += 1
    subfolders = temp
    print(subfolders)
```

```python
missing_L = []
expected_exts = {'.params', '.pickle', '.png', '.rois'}

rows=[]

for fov in sorted(fovs):
    fov_path = os.path.join(root, fov)
    subfolders = os.listdir(fov_path)
    fov_issues = []

    if len(subfolders) != 14:
        fov_issues.append(f'  expected 14 subfolders, got {len(subfolders)} -> {sorted(subfolders)}')

    
    for sub in subfolders:
        sub_issues = []
        
        if not re.match(r'^\d{6}$', sub):
            fov_issues.append(f'  {sub}: main date folder - date is wrong')
            continue
        
        sub_path = os.path.join(fov_path, sub)
        files = [f for f in os.listdir(sub_path) if f != 'Thumbs.db']
        
        if len(files) != 4:
            sub_issues.append(f'expected 4 files, got {len(files)} -> {files}')
        
        exts = set(os.path.splitext(f)[1] for f in files)
        missing = expected_exts - exts
        extra = exts - expected_exts
        if missing:
            sub_issues.append(f'missing {missing}')
        
        for f in files:
            stem = os.path.splitext(f)[0]
            m = re.match(r'^JL.+_(\d{6})_imaging_data_L\d+$', stem)
            if not m:
                if re.match(r'^JL.+_\d{6}_imaging_data$', stem):
                    missing_L.append((fov, sub_path, f))
                    sub_issues.append(f'missing layer number: {f}')
                else:
                    sub_issues.append(f'wrong filename: {f}')
            elif m.group(1) != sub:
                sub_issues.append(f'date mismatch: folder is {sub} but file has {m.group(1)} ({f})')
        
        if sub_issues:
            fov_issues.append(f'  {sub}: ' + ' | '.join(sub_issues))
    
    if fov_issues:
        print(f'\n{fov} ({len(fov_issues)} issues):')
        for issue in fov_issues:
            print(issue)
    else:
        print(f'{fov}: ok')
print('\ndone')
```

```python
# check motion_corrected folders
mc_root = r"Z:\People\JenLi\Imaging_Data\L23_L5_structural_imaging\motion_corrected"

expected_dates = {
    'JL117': ['250211','250212','250213','250214','250215','250216','250217','250218','250219','250220','250221','250222','250224','250225'],
    'JL122': ['250402','250403','250404','250405','250406','250407','250408','250409','250410','250411','250412','250413','250414','250415'],
    'JL123': ['250402','250403','250404','250405','250406','250407','250408','250409','250410','250411','250412','250413','250414','250415'],
}

for jl in sorted(os.listdir(mc_root)):
    jl_path = os.path.join(mc_root, jl)
    if not os.path.isdir(jl_path) or not jl.startswith('JL'):
        continue

    expected = expected_dates.get(jl, [])

    # collect all FOV names across all date folders
    all_fovs = set()
    for d in expected:
        d_path = os.path.join(jl_path, d)
        if not os.path.isdir(d_path):
            continue
        for fov in os.listdir(d_path):
            if re.match(r'^FOV\d+$', fov):
                all_fovs.add(fov)

    for fov in sorted(all_fovs):
        print(f'\n{jl}, {fov}')

        # which dates have this FOV folder
        dates_present = [d for d in expected if os.path.isdir(os.path.join(jl_path, d, fov))]
        missing_dates = [d for d in expected if d not in dates_present]
        print(f'  date folders:     {len(dates_present)}/14', end='')
        print(f'  -- missing: {missing_dates}' if missing_dates else '')

        # which dates have GreenCh
        has_green = []
        missing_green = []
        for d in dates_present:
            fov_path = os.path.join(jl_path, d, fov)
            ch_folders = os.listdir(fov_path)
            if any('Green' in c or 'Red' in c for c in ch_folders):
                has_green.append(d)
            else:
                missing_green.append(d)
        print(f'  Ch folder present:  {len(has_green)}/14', end='')
        print(f'  -- missing: {missing_green}' if missing_green else '')

        # which GreenCh have XYZAligned.tif
        has_xyz = []
        missing_xyz = []
        for d in has_green:
            fov_path = os.path.join(jl_path, d, fov)
            ch_folders = os.listdir(fov_path)
            green = next((c for c in ch_folders if 'Green' in c or 'Red' in c), None)
            green_path = os.path.join(fov_path, green)
            files = [f for f in os.listdir(green_path) if f != 'Thumbs.db']
            if 'XYZAligned.tif' in files:
                has_xyz.append(d)
            else:
                missing_xyz.append(d)
        print(f'  XYZAligned.tif:   {len(has_xyz)}/14', end='')
        print(f'  -- missing: {missing_xyz}' if missing_xyz else '')

print('\ndone')
```
