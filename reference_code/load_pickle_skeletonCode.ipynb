```python
# Spine Add/Elim

all_days = {}
for day in range(1,14):
    out = utils.add_elim_spines_days(df, day, day+1)
    utils.save_all_days(all_days, out)
```

```python
# dend.plot.new_elim_bydend

# New/eliminated spines per µm, per dendrite

x = list(range(2, 15))
for dendrite, ts in all_days.items():
    plt.figure()
    plt.plot(x, ts["elim_per"], marker="o", label="eliminated spines per µm")
    plt.plot(x, ts["new_per"], marker="o", linestyle="--", label="new spines per µm")

    plt.xlabel("day")
    plt.ylabel("spines per µm")
    plt.title(dendrite)
    plt.xticks(x)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
# Fraction of new/eliminated spines per µm, per dendrite

x = list(range(2, 15))
for dendrite, ts in all_days.items():
    plt.figure()
    new_values = np.array(ts["new_per"])
    elim_values = np.array(ts["elim_per"])

    plt.bar(x, new_values, label="new spines per µm")
    plt.bar(x, -elim_values, label="eliminated spines per µm")

    plt.axhline(0, linewidth=1)

    plt.xlabel("day")
    plt.ylabel("spines per µm")
    plt.title(dendrite)
    plt.xticks(x)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
# Spine turnover per µm, per dendrite

x = list(range(2, 15))
for dendrite, ts in all_days.items():
    turnover_per = np.array(ts["elim_per"]) + np.array(ts["new_per"])

    plt.figure()
    plt.plot(x, turnover_per, marker="o")
    plt.axhline(0, linewidth=1)

    plt.xlabel("day")
    plt.ylabel("turnover per µm")
    plt.title(dendrite)
    plt.xticks(x)
    plt.tight_layout()
    plt.show()
```
